```python
total_spent_mask = df['Total Spent'].isna() & df[['Quantity','Price Per Unit']].notna().all(axis=1)
df.loc[total_spent_mask,'Total Spent'] = df['Price Per Unit'] * df['Quantity']
```

```python
df['Price Per Unit'] = pd.to_numeric(df['Price Per Unit'], errors='coerce')
df['Quantity'] = pd.to_numeric(df['Quantity'], errors='coerce').astype('float')
df['Total Spent'] = pd.to_numeric(df['Total Spent'], errors='coerce')

#Filling Price per unit column by using map of item and its prices
price_map = df.groupby('Item')['Price Per Unit'].median().to_dict()
unit_price_mask = df['Price Per Unit'].isna()
df.loc[unit_price_mask, 'Price Per Unit'] = df.loc[unit_price_mask, 'Item'].map(price_map)

#Finally filling up the Quantity and Total Spent one more time since we filled up Price Per Unit more
quantity_null_mask = (df['Quantity'].isna() & df['Total Spent'].notna() & df['Price Per Unit'].notna() & (df['Price Per Unit'] != 0))
df.loc[quantity_null_mask, 'Quantity'] = (df.loc[quantity_null_mask, 'Total Spent'] / df.loc[quantity_null_mask, 'Price Per Unit'])
df['Quantity'] = df['Quantity'].round().astype('Int64')
total_spent_null_mask = (df['Total Spent'].isna() & df[['Price Per Unit', 'Quantity']].notna().all(axis=1))
df.loc[total_spent_null_mask, 'Total Spent'] = (df.loc[total_spent_null_mask, 'Quantity'] * df.loc[total_spent_null_mask, 'Price Per Unit'])
```

```python
#filling item column based on prices using probability so i wont create the biased data
no_nan_items = df.dropna(subset='Item')
item_percentages = no_nan_items.groupby('Price Per Unit')['Item'].value_counts(normalize=True)
price_item_map = {}
for index in item_percentages.index.get_level_values(0).unique():
    items = item_percentages[index].index.tolist()
    percentages = item_percentages[index].values.tolist()

    price_item_map[index] = {
        'items':items,
        'probability':percentages
    }


def fill_missing(per_unit_price):
    if per_unit_price in price_item_map:
        return np.random.choice(a=price_item_map[per_unit_price]['items'], p=price_item_map[per_unit_price]['probability'])
    else:
        return np.nan
        

#applying function to each row in item column:
item_nan_mask = df['Item'].isna()
df.loc[item_nan_mask, 'Item'] = df.loc[item_nan_mask, 'Price Per Unit'].map(fill_missing)
```

```python
df.replace(r'(?i)(\berror\b|\bunknown\b)', np.nan, regex=True, inplace=True)
```

```python
#Droping rows with na values that cant be filled
df = df.dropna(how='any', subset=['Quantity', 'Price Per Unit', 'Total Spent'])
#Finally filling up all NA values with 'Unknown' and checking for any errors with filling up
df = df.fillna('Unknown')
(df.isna().mean() * 100).round(2)
```
